```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
```

```python
def plot_loss(train_loss, val_loss, file_name, y_axis_title="Loss"):
    steps = np.arange(len(train_loss))
    plt.figure(figsize=(4, 3), dpi=300)

    viridis = plt.cm.viridis
    plt.plot(steps, train_loss, label="Training Loss", color=viridis(0.2), linewidth=1.5)
    
    plt.plot(steps, val_loss, label="Validation Loss", color=viridis(0.7), linewidth=1.5)
    
    plt.xlabel("Training Step")
    plt.ylabel(y_axis_title)
    # plt.title("Loss Curve")
    plt.legend()
    plt.grid(True, linestyle="--", linewidth=0.5, alpha=0.6)
    
    plt.tight_layout()
    plt.savefig(file_name, dpi=300)
    plt.show()
```

```python
# VQ VAE Total
vq_train_loss_path = "data/loss/vqvae_train_loss.csv"
vq_val_loss_path = "data/loss/vqvae_val_loss.csv"

vqvae_train_df =  pd.read_csv(vq_train_loss_path)
vqvae_val_df = pd.read_csv(vq_val_loss_path)

train_loss = vqvae_train_df["debug--VQVAE_CSL_ALL_ORIG_1GPU - total/train"]
val_loss = vqvae_val_df["debug--VQVAE_CSL_ALL_ORIG_1GPU - total/val"]

plot_loss(train_loss, val_loss, "figs/vqvae_total_loss_curve.png", "Total Loss")
```
